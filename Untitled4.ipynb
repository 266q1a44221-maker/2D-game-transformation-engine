```python
from IPython.display import HTML, display

html = """
<!DOCTYPE html>
<html>
<head>
<title>2D Game Transformation Engine</title>

<style>
body {
    background: #111827;
    color: white;
    font-family: Arial;
    text-align: center;
}

canvas {
    background: #1f2937;
    border: 2px solid cyan;
    max-width: 95%;
}

button {
    padding: 10px;
    margin: 5px;
    background: #06b6d4;
    color: white;
    border: none;
    border-radius: 6px;
    font-weight: bold;
}

.box {
    margin: 15px auto;
    padding: 15px;
    max-width: 700px;
    background: #1e293b;
    border-radius: 10px;
}

pre {
    text-align: left;
    display: inline-block;
    margin: 10px;
}
</style>
</head>

<body>

<h1>2D Game Transformation Engine</h1>

<canvas id="canvas" width="700" height="400"></canvas>

<div class="box">

<h2>Controls</h2>

<button onclick="moveLeft()">Left</button>
<button onclick="moveRight()">Right</button>
<button onclick="moveUp()">Up</button>
<button onclick="moveDown()">Down</button>

<br>

<button onclick="rotateLeft()">Rotate Left</button>
<button onclick="rotateRight()">Rotate Right</button>

<br>

<button onclick="scaleUp()">Increase Size</button>
<button onclick="scaleDown()">Decrease Size</button>

<br>

<button onclick="resetGame()">Reset</button>

</div>

<div class="box">

<h2>Transformation</h2>

<p>X: <span id="x">350</span></p>
<p>Y: <span id="y">200</span></p>
<p>Angle: <span id="angle">0</span> degrees</p>
<p>Scale: <span id="scale">1.00</span></p>

</div>

<div class="box">

<h2>Matrices</h2>

<h3>Translation Matrix T</h3>
<pre id="T"></pre>

<h3>Rotation Matrix R</h3>
<pre id="R"></pre>

<h3>Scaling Matrix S</h3>
<pre id="S"></pre>

<h3>Total = T x R x S</h3>
<pre id="TOTAL"></pre>

</div>

<script>

var canvas = document.getElementById("canvas");
var ctx = canvas.getContext("2d");

var x = 350;
var y = 200;
var angle = 0;
var scale = 1;


/* Translation */

function T(tx, ty) {
    return [
        [1, 0, tx],
        [0, 1, ty],
        [0, 0, 1]
    ];
}


/* Rotation */

function R(a) {

    var r = a * Math.PI / 180;

    var c = Math.cos(r);
    var s = Math.sin(r);

    return [
        [c, -s, 0],
        [s, c, 0],
        [0, 0, 1]
    ];
}


/* Scaling */

function S(s) {
    return [
        [s, 0, 0],
        [0, s, 0],
        [0, 0, 1]
    ];
}


/* Matrix multiplication */

function multiply(A, B) {

    var C = [
        [0,0,0],
        [0,0,0],
        [0,0,0]
    ];

    var i;
    var j;
    var k;

    for (i = 0; i < 3; i++) {

        for (j = 0; j < 3; j++) {

            for (k = 0; k < 3; k++) {

                C[i][j] =
                    C[i][j] +
                    A[i][k] * B[k][j];

            }
        }
    }

    return C;
}


/* Show matrix */

function show(M) {

    var text = "";

    for (var i = 0; i < 3; i++) {

        text += "[ ";

        for (var j = 0; j < 3; j++) {

            text += M[i][j].toFixed(2) + " ";

        }

        text += "]\\n";
    }

    return text;
}


/* Update matrices */

function updateMatrices() {

    var tx = x - 350;
    var ty = y - 200;

    var TM = T(tx, ty);
    var RM = R(angle);
    var SM = S(scale);

    var RS = multiply(RM, SM);

    var TOTAL = multiply(TM, RS);

    document.getElementById("T").textContent = show(TM);

    document.getElementById("R").textContent = show(RM);

    document.getElementById("S").textContent = show(SM);

    document.getElementById("TOTAL").textContent = show(TOTAL);

    document.getElementById("x").textContent = x;

    document.getElementById("y").textContent = y;

    document.getElementById("angle").textContent = angle;

    document.getElementById("scale").textContent =
        scale.toFixed(2);
}


/* Draw grid */

function grid() {

    ctx.strokeStyle = "#334155";

    for (var i = 0; i <= 700; i += 35) {

        ctx.beginPath();
        ctx.moveTo(i, 0);
        ctx.lineTo(i, 400);
        ctx.stroke();
    }

    for (var j = 0; j <= 400; j += 35) {

        ctx.beginPath();
        ctx.moveTo(0, j);
        ctx.lineTo(700, j);
        ctx.stroke();
    }
}


/* Draw triangle */

function drawTriangle() {

    ctx.save();

    ctx.translate(x, y);

    ctx.rotate(angle * Math.PI / 180);

    ctx.scale(scale, scale);

    ctx.fillStyle = "#00ffcc";

    ctx.beginPath();

    ctx.moveTo(0, -50);

    ctx.lineTo(50, 50);

    ctx.lineTo(-50, 50);

    ctx.closePath();

    ctx.fill();

    ctx.restore();
}


/* Draw everything */

function draw() {

    ctx.clearRect(0, 0, 700, 400);

    grid();

    drawTriangle();

    updateMatrices();
}


/* Movement */

function moveLeft() {
    x -= 15;
    draw();
}

function moveRight() {
    x += 15;
    draw();
}

function moveUp() {
    y -= 15;
    draw();
}

function moveDown() {
    y += 15;
    draw();
}


/* Rotation */

function rotateLeft() {
    angle -= 15;
    draw();
}

function rotateRight() {
    angle += 15;
    draw();
}


/* Scaling */

function scaleUp() {

    scale += 0.1;

    if (scale > 2) {
        scale = 2;
    }

    draw();
}


function scaleDown() {

    scale -= 0.1;

    if (scale < 0.5) {
        scale = 0.5;
    }

    draw();
}


/* Reset */

function resetGame() {

    x = 350;
    y = 200;
    angle = 0;
    scale = 1;

    draw();
}


/* Start */

draw();

</script>

</body>
</html>
"""

display(HTML(html))
```
